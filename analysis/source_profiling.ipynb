## First Look

### Raw Source

```python
API_URL = "https://stats.tennismylife.org/api/data-files"

response = requests.get(API_URL).json()
```

### Filtering Relevant Files

ATP (`<year>.csv`) and WTA (`<year>_wta.csv`) only — see #14/#17 for why Challenger, Qualifying, amateur, backup, and ongoing files are excluded.

```python
pattern = re.compile(r"^\d{4}(_wta)?\.csv$")

matches = [item["name"] for item in response["files"] if pattern.match(item["name"])]
matches
```

```python
atp_files = [
    item
    for item in response["files"]
    if item["name"] in matches and not item["name"].endswith("_wta.csv")
]
wta_files = [
    item
    for item in response["files"]
    if item["name"] in matches and item["name"].endswith("_wta.csv")
]

atp_by_name = {item["name"]: item["url"] for item in atp_files}
wta_by_name = {item["name"]: item["url"] for item in wta_files}

print(f"ATP files: {len(atp_files)}, WTA files: {len(wta_files)}")
```

### Basic Exploration

I'm taking the files from 2025 to understand the data, since it's fairly recent and covers a full year.

```python
atp_2025 = pd.read_csv(atp_by_name["2025.csv"])
wta_2025 = pd.read_csv(wta_by_name["2025_wta.csv"])
```

#### 3. What does `tourney_date` mean, and what's the distribution of values?

My takeaways:

Norrie vs. Pieczkowski (POL vs GBR): `w_1stIn=62` exactly equals `l_svpt=62` — looks like a value misassigned between winner/loser, not just a rounding slip. #### 12. Do categorical columns match the values documented in the TML-Database dictionary?

Dictionary source: the site’s own [tournament-database page](https://stats.tennismylife.org/tennis-match-database), pasted in full by Ana after my first WebFetch summary silently dropped the `indoor` row — cross-checked against the [Tennismylife/TML-Database README](https://github.com/Tennismylife/TML-Database), which agrees on every field.

Caveat on `entry`: the site only gives examples ("e.g., Q, WC"), not an exhaustive list — unlike GitHub, which states `Q, WC, LL, SE` as complete. So an entry-code "violation" below isn’t necessarily a data error; it could be a legitimate code neither source bothered to document (e.g. "PR" for protected ranking).

My takeaways:
- `indoor`: documented as `Yes/No`, but the real values are `I`/`O` — almost certainly Indoor/Outdoor codes, not a boolean. Needs a mapping decision in the raw/bronze schema (e.g. `I` → true), not a straight boolean cast.
- `round`: `RR` (round robin) undocumented but expected — already known from question 3.
- `tourney_level`: actual values are `250`, `500`, `M` — the source uses ATP’s real tier naming, not the documented `G/A/D/F` scheme. Documentation is stale here, not the data.
- `entry`: `PR` (protected ranking), `ALT` (alternate) — real, legitimate codes just not documented (consistent with the site’s own "e.g." caveat). `NG` (loser_entry, 3 rows, all ATP 250 grass events) — meaning not confirmed; scores are normal (no RET/WO), so not a withdrawal marker.
- `draw_size`: extra values (18, 28, 48, 56, 96, 2) all trace back to known non-standard formats already explored — United Cup, 28-draws, Madrid-style combined events, Laver Cup.

```python
expected = {
    "surface": {"Hard", "Clay", "Grass", "Carpet"},
    "draw_size": {128, 64, 32, 16, 8, 4},
    "tourney_level": {"G", "A", "D", "F"},
    "indoor": {"Yes", "No"},
    "winner_entry": {"Q", "WC", "LL", "SE"},
    "loser_entry": {"Q", "WC", "LL", "SE"},
    "winner_hand": {"R", "L"},
    "loser_hand": {"R", "L"},
    "round": {"R128", "R64", "R32", "R16", "QF", "SF", "F"},
    "best_of": {3, 5},
}

atp_dict_violations = {
    col: sorted(set(atp_2025[col].dropna().unique()) - allowed, key=str)
    for col, allowed in expected.items()
}
{col: vals for col, vals in atp_dict_violations.items() if vals}
```

## Historical Consistency

### ATP

My takeaways:
- Column structure identical across 1970/1995/2010/2025 — safe to assume a stable ATP schema.

### 3. Is `tourney_id` unique to one tournament? Is `match_num` unique within it?

My takeaways:
- **`tourney_id` collision, real bug**: `2026-416` holds 1 row for Munich (13 Apr) and the entire 95-match Rome Masters draw (6-17 May) — different city, different dates, same id. 0 such collisions in ATP/WTA 2025 or ATP 2026 otherwise — looks like an isolated glitch in this one live-season id, not systemic.
- **`tourney_id` sharing, but structural, not a bug**: WTA 2025 has 2 ids each covering several distinct group-stage "tourney_name"s — `2025-W-FC-2025-POS` spans Billie Jean King Cup Play-offs Groups A-G, `2025-W-FC-2025-QUA` spans Qualifying Round Groups A-F. Consistent with the known BJK Cup team-event pattern from #14 (multiple dates per tourney_id) — the source seems to key BJK Cup by *phase*, not by group, so grouping there is intentional, unlike Munich/Rome.
- **`match_num` duplicate, real bug**: `2026-422` (Cincinnati Masters) has two distinct real matches both numbered 79 (R32) — numbering skips 80 and repeats 79. 0 such real duplicates found elsewhere (ATP/WTA 2025, WTA 2026) once null `match_num` rows are excluded from the check (see the fixed helper below — first pass wrongly counted 489 null-vs-null "duplicates" in ATP 2025 before the fix).
- Net: both `tourney_id` and `match_num` glitches found so far look like rare, isolated data bugs rather than a recurring pattern, plus one structural case (BJK Cup) that isn't a bug at all. Worth sampling a few more years before concluding it's *always* rare.

```python
def tourney_id_collisions(df):
    """Return rows where the same tourney_id maps to more than one tourney_name."""
    name_counts = df.groupby("tourney_id")["tourney_name"].nunique()
    collided_ids = name_counts[name_counts > 1].index
    return (
        df[df["tourney_id"].isin(collided_ids)][
            ["tourney_id", "tourney_name", "tourney_date"]
        ]
        .drop_duplicates()
        .sort_values("tourney_id")
    )


pd.concat(
    {
        "ATP 2026": tourney_id_collisions(atp_2026),
        "WTA 2026": tourney_id_collisions(wta_2026),
    },
    names=["tour"],
)
```

```python
def match_num_real_duplicates(df):
    """Same (tourney_id, match_num) but different players -- a numbering bug, not a row duplicate."""
    populated = df.dropna(subset=["match_num"])
    dupes = populated[
        populated.duplicated(subset=["tourney_id", "match_num"], keep=False)
    ]
    return dupes[
        ~dupes.duplicated(
            subset=["tourney_id", "match_num", "winner_name", "loser_name"], keep=False
        )
    ]


pd.concat(
    {
        "ATP 2026": match_num_real_duplicates(atp_2026),
        "WTA 2026": match_num_real_duplicates(wta_2026),
    },
    names=["tour"],
)[["tourney_id", "tourney_name", "round", "match_num", "winner_name", "loser_name"]]
```
