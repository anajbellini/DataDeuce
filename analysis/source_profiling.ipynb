# Data Deuce: Source Profiling

## About the Source

My plan is to use [TennisMyLife's database](https://stats.tennismylife.org/tennis-match-database) available via API.

## Libraries

```python
import re

import pandas as pd
import requests
```

## Raw Source

```python
API_URL = "https://stats.tennismylife.org/api/data-files"

response = requests.get(API_URL).json()
```

## Filtering Relevant Files

ATP (`<year>.csv`) and WTA (`<year>_wta.csv`) only — see #14/#17 for why Challenger, Qualifying, amateur, backup, and ongoing files are excluded.

```python
pattern = re.compile(r"^\d{4}(_wta)?\.csv$")

matches = [item["name"] for item in response["files"] if pattern.match(item["name"])]
matches
```

## Basic Exploration

I'm taking the files from 2025 to understand the data, since it's fairly recent and covers a full year.

```python
target_names = {"2025.csv", "2025_wta.csv"}
urls = {
    item["name"]: item["url"]
    for item in response["files"]
    if item["name"] in target_names
}

atp_2025 = pd.read_csv(urls["2025.csv"])
wta_2025 = pd.read_csv(urls["2025_wta.csv"])
```

```python
print("ATP 2025:", atp_2025.shape)
print("WTA 2025:", wta_2025.shape)
```
