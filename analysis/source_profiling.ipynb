# Data Deuce: Source Profiling

## About the Source

My plan is to use [TennisMyLife's database](https://stats.tennismylife.org/tennis-match-database) available via API.

## Libraries

```python
import re

import pandas as pd
import requests
```

## First Look

### Raw Source

```python
API_URL = "https://stats.tennismylife.org/api/data-files"

response = requests.get(API_URL).json()
```

### Filtering Relevant Files

ATP (`<year>.csv`) and WTA (`<year>_wta.csv`) only — see #14/#17 for why Challenger, Qualifying, amateur, backup, and ongoing files are excluded.

```python
pattern = re.compile(r"^\d{4}(_wta)?\.csv$")

matches = [item["name"] for item in response["files"] if pattern.match(item["name"])]
matches
```

```python
atp_files = [
    item
    for item in response["files"]
    if item["name"] in matches and not item["name"].endswith("_wta.csv")
]
wta_files = [
    item
    for item in response["files"]
    if item["name"] in matches and item["name"].endswith("_wta.csv")
]

atp_by_name = {item["name"]: item["url"] for item in atp_files}
wta_by_name = {item["name"]: item["url"] for item in wta_files}

print(f"ATP files: {len(atp_files)}, WTA files: {len(wta_files)}")
```

### Basic Exploration

I'm taking the files from 2025 to understand the data, since it's fairly recent and covers a full year.

```python
atp_2025 = pd.read_csv(atp_by_name["2025.csv"])
wta_2025 = pd.read_csv(wta_by_name["2025_wta.csv"])
```

## Deep Dive

### ATP

#### 1. What's the available date range?

My takeaways:
- `2024-12-29` to `2025-12-22` — not a clean calendar year, since United Cup (pre-season) starts in late December of the prior year.

```python
atp_2025_dates = pd.to_datetime(atp_2025["tourney_date"], format="%Y%m%d")
atp_2025_dates.min(), atp_2025_dates.max()
```

#### 2. Is there a timestamp? If so, what timezone?

My takeaways:
- No timestamp, no timezone — only `tourney_date` (`int64`, `YYYYMMDD`), tz-naive.

```python
date_like_cols = atp_2025.filter(regex="date|time|tz", axis=1).columns.tolist()
date_like_cols, atp_2025["tourney_date"].dtype, atp_2025_dates.dt.tz
```

#### 3. What does `tourney_date` mean, and what's the distribution of values?

My takeaways:
- `tourney_date` = tournament-level anchor, constant per `tourney_id` — not a per-match date.
- Confirmed as **start of tournament week** against Australian Open 2025 (`20250113`, the real Monday).
- No per-match date field exists anywhere — this source only gives week-level granularity.
- Exceptions: **WTA Finals** (round-robin) and **BJK Cup** (team event) span multiple dates per `tourney_id`.

```python
atp_2025.groupby("tourney_id")["tourney_date"].nunique().value_counts()
```

#### 4. Does match count per tournament match `draw_size`?

My takeaways:
- 110/197 tourneys mismatched — mostly round-robin (ATP/Next Gen Finals), team/exhibition formats (Laver Cup, United Cup) where `draw_size` isn't bracket slots, and Davis Cup split into one `tourney_id` per tie-match (`draw_size=4`, single match logged).
- Dubai's lone non-format mismatch traced to a duplicate row — see question 6.

```python
atp_match_counts = atp_2025.groupby("tourney_id").agg(
    tourney_name=("tourney_name", "first"),
    draw_size=("draw_size", "first"),
    actual_matches=("match_num", "size"),
)
atp_match_counts["expected_matches"] = atp_match_counts["draw_size"] - 1
atp_match_counts["diff"] = (
    atp_match_counts["actual_matches"] - atp_match_counts["expected_matches"]
)

atp_match_counts[atp_match_counts["diff"] != 0]
```

```python
dubai = atp_2025[atp_2025["tourney_id"] == "2025-495"]
dubai["round"].value_counts()
```

```python
dubai[dubai.duplicated(subset=["winner_name", "loser_name", "round"], keep=False)][
    ["round", "match_num", "winner_name", "loser_name", "score"]
]
```

```python
dubai[dubai["score"].str.contains(r"[^0-9\-\s()]", na=False)][
    ["round", "match_num", "winner_name", "loser_name", "score"]
]
```

#### 5. How do non-standard outcomes (retirement, walkover, default) show up in `score`?

My takeaways:
- 102 distinct patterns, all either RET (~100, partial set score + retirement) or W/O (22, clean). No DEF or ABD found in 2025.

```python
atp_2025[atp_2025["score"].str.contains(r"[^0-9\-\s()]", na=False)][
    "score"
].value_counts()
```

#### 6. Are there duplicate match rows?

My takeaways:
- 0 full-row duplicates; 1 business-key duplicate (`tourney_id` + players + `round`) — the Nardi/Fucsovics case from Dubai (question 4), isolated, not systemic.

```python
atp_2025.duplicated().sum()
```

```python
atp_2025.duplicated(subset=["tourney_id", "winner_name", "loser_name", "round"]).sum()
```

#### 7. Of the tournaments that don't match `draw_size`, which aren't explained by a known special format (Finals/Cup)?

```python
atp_mismatches = atp_match_counts[atp_match_counts["diff"] != 0]
known_format = atp_mismatches["tourney_name"].str.contains(
    "Finals|Cup", case=False, na=False
)

atp_mismatches[~known_format]
```

#### 8. How do bye cases show up in the data?

```python
(
    atp_2025["round"].unique(),
    atp_2025[atp_2025["score"].str.contains("bye", case=False, na=False)],
)
```

My takeaways:
- Same as ATP: no timestamp, no timezone — only `tourney_date` (`int64`, `YYYYMMDD`), tz-naive.

```python
date_like_cols = wta_2025.filter(regex="date|time|tz", axis=1).columns.tolist()
date_like_cols, wta_2025["tourney_date"].dtype, wta_2025_dates.dt.tz
```

#### 3. What does `tourney_date` mean, and what's the distribution of values?

My takeaways:
- Same as ATP: constant per `tourney_id` for 55/57 tournaments — start-of-week date.
- Exceptions: **WTA Finals** (2 dates) and **BJK Cup** (8 dates) — legitimate, not a bug.
- No per-match date field anywhere in this source.

```python
wta_2025.groupby("tourney_id")["tourney_date"].nunique().value_counts()
```

```python
wta_date_counts = wta_2025.groupby("tourney_id")["tourney_date"].nunique()
anomalous_tourney_ids = wta_date_counts[wta_date_counts > 1].index

wta_2025[wta_2025["tourney_id"].isin(anomalous_tourney_ids)][
    ["tourney_id", "tourney_name", "tourney_date", "round", "match_num"]
].sort_values(["tourney_id", "tourney_date"])
```

#### 4. Does match count per tournament match `draw_size`?

My takeaways:
- 7/57 tourneys mismatched — United Cup and WTA Finals/BJK Cup are known non-standard formats; Washington and Linz traced to the `draw_size` bracket-vs-entrant inconsistency — see questions 7-8.

```python
wta_match_counts = wta_2025.groupby("tourney_id").agg(
    tourney_name=("tourney_name", "first"),
    draw_size=("draw_size", "first"),
    actual_matches=("match_num", "size"),
)
wta_match_counts["expected_matches"] = wta_match_counts["draw_size"] - 1
wta_match_counts["diff"] = (
    wta_match_counts["actual_matches"] - wta_match_counts["expected_matches"]
)

wta_match_counts[wta_match_counts["diff"] != 0]
```

```python
wta_2025[wta_2025["score"].str.contains(r"[^0-9\-\s()]", na=False)][
    "score"
].value_counts()
```

#### 6. Are there duplicate match rows?

My takeaways:

#### 7. Of the tournaments that don't match `draw_size`, which aren't explained by a known special format (Finals/Cup)?

My takeaways:
- Washington and Linz: both short exactly 4 matches, only in `R32` — no `score` anomalies (RET/W/O).
- **Confirmed externally**: both are real **28-draw WTA events** ([Washington](https://www.wtatennis.com/news/4315457/washington-2025-draws-dates-prize-money-and-what-you-need-to-know), [Linz](https://en.wikipedia.org/wiki/2025_Upper_Austria_Ladies_Linz)) — 4 byes for seeds 1-4 is standard for this format, not an anomaly.
- Root cause: `draw_size` mixes two meanings — administrative bracket size (32, here) vs. true entrant count (96 for Madrid, question 8). **`draw_size - 1` isn't a reliable row-count check**; validating round structure is.

```python
wta_mismatches = wta_match_counts[wta_match_counts["diff"] != 0]
known_format = wta_mismatches["tourney_name"].str.contains(
    "Finals|Cup", case=False, na=False
)

wta_mismatches[~known_format]
```
